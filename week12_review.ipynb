## **4. 새롭게 알게된 점/ 배운점**

- 앙상블 학습(Ensemble Learning): 여러 개의 모델을 결합하여 예측 성능을 높이는 학습 방법
- Bootstrap Sampling: 복원추출을 통해 원본 데이터와 다른 학습 샘플을 생성하는 방법
- Random Forest: 여러 결정 트리를 학습시키고 예측 결과를 종합하여 과적합을 줄이고 일반화 성능을 높이는 모델

- Extra Trees(Extremely Randomized Trees):
  - 기본적으로 100개의 결정 트리 사용
  - Bootstrap Sampling 없이 전체 훈련 데이터 사용
  - 노드 분할 시 최적의 분할점을 찾는 대신 무작위 분할점 활용
  
- 특성 중요도: feature_importances_를 통해 모델 예측에 중요한 변수를 파악할 수 있음
